```python
sig = mat["Signal_1"][0][0][1][0][0][0][:, 0]
# px.line(sig)
plt.plot(sig)
len(sig) / 20480
```

```python
rot_freq = 15.2
# rot_freq = 937 / 60
bpfo = 8.1765 * rot_freq
bpfi = 10.824 * rot_freq

sig = np.abs(scipy.signal.hilbert(sig))

fft = np.abs(np.fft.rfft(sig[: 20480 * 10]))
fftfreq = np.fft.rfftfreq(len(sig[: 20480 * 10]), 1 / 20480)

f = px.line(x=fftfreq, y=fft)

for i in range(5):
    f.add_vline(x=bpfo * (i + 1), line_dash="dash", line_color="green")
    f.add_vline(x=bpfi * (i + 1), line_dash="dash", line_color="blue")
f.show()
```
